# **Exploratory Data Analyse**

```python
## Importation des données
import pandas as pd
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
os.chdir('../data')
```

```python
clean_df=pd.read_json('raw/IntroMLops-1.json')
```

```python
clean_df = clean_df[clean_df['temperature'] >= 30.]
clean_df ['ratio'] = clean_df['intensity'] / clean_df['humidity']
```

### **Test de corelation**

```python
correlation_matrix=clean_df[['temperature', 'intensity', 'humidity']].corr()
print("Correlation Matrix")
correlation_matrix
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
```

```python
fig,ax=plt.subplots(1, 3,figsize=(18, 5))
for i,column in enumerate(['temperature', 'intensity', 'humidity']):
    sns.histplot(clean_df[column], kde=True, label=column, bins=30,ax=ax[i])
    ax[i].set_title(f'Distribution of {column}')
    ax[i].set_xlabel('Value')
    ax[i].set_ylabel('Frequency')
plt.legend()
plt.show()
```

```python
plot_pair = sns.pairplot(clean_df[['temperature', 'intensity', 'humidity']])
plot_pair.fig.suptitle('Pairwise Relationships')
```

```python
boxplot_fig, ax = plt.subplots(1,3,figsize=(10, 6))
for i, column in enumerate(['temperature', 'intensity', 'humidity']):
    sns.boxplot(data=clean_df[[column]], ax=ax[i])
    ax[i].set_title(f'Boxplot of {column}')
    ax[i].set_ylabel('Value')
    ax[i].set_xlabel(column)
    plt.show()
```
